```python
import pulp
```

```python
# create problem
prob = pulp.LpProblem("generations", pulp.LpMaximize)

# variables
P = pulp.LpVariable("p",lowBound=0,cat=pulp.LpInteger)
x = pulp.LpVariable("x",lowBound=0,upBound=9,cat=pulp.LpInteger)
y = pulp.LpVariable("y",lowBound=0,upBound=9,cat=pulp.LpInteger)

Pminus1 = pulp.LpVariable("pm1",lowBound=0,cat=pulp.LpInteger)
a = pulp.LpVariable("a",lowBound=0,upBound=9,cat=pulp.LpInteger)
b = pulp.LpVariable("b",lowBound=0,upBound=9,cat=pulp.LpInteger)

F = pulp.LpVariable("f",lowBound=0,cat=pulp.LpInteger)

GP = pulp.LpVariable("gp",lowBound=0,cat=pulp.LpInteger)
GPminus1 = pulp.LpVariable("gpm1",lowBound=0,cat=pulp.LpInteger)

# dummy objective function
prob += 0

# constraints
prob+= P == 10*x+y
prob+= F == 10*y+x
prob+= GP-P == P-F
prob+= P-Pminus1 == 1
prob+= GP-GPminus1 == 1
prob+= Pminus1 == 10*a+b
prob+= GPminus1 == 10*b+a
```

```python
prob.solve()
```

```python
print("(F,P,GP)=",(pulp.value(F),pulp.value(P),pulp.value(GP)))
```

```python
print("P-1=",pulp.value(Pminus1),"GP-1=",pulp.value(GPminus1))
```
